## Date-Time Cleaning & Feature Engineering

## Create Patient Age Groups

```python
def age_group(age):

    if pd.isna(age):
        return "Unknown"

    elif age <= 25:
        return "Young Adult"

    elif age <= 40:
        return "Adult"

    elif age <= 60:
        return "Middle Age"

    else:
        return "Senior"
age_group
```

```python
patient_vitals["Age_Group"] = (
    patient_vitals["Age"].apply(age_group)
)
print(
    patient_vitals["Age_Group"].value_counts()
)
```

## Standardize Categorical Columns

```python
categorical_columns = [
    "Specialization",
    "Consultation_Status",
    "Payment_Status",
    "Health_Status"
]
for column in categorical_columns:

    if column in teleconsultation.columns:
        teleconsultation[column] = (
            teleconsultation[column]
            .astype("string")
            .str.strip()
            .str.title()
        )

    if column in sales_data.columns:
        sales_data[column] = (
            sales_data[column]
            .astype("string")
            .str.strip()
            .str.title()
        )

    if column in patient_vitals.columns:
        patient_vitals[column] = (
            patient_vitals[column]
            .astype("string")
            .str.strip()
            .str.title()
        )
print(categorical_columns)
```

## Custom Health Risk Classification Function

```python
def categorize_health_risk(row):

    indicators = 0

    if row["SpO2_Percentage"] < 94:
        indicators += 1

    if row["Heart_Rate_bpm"] > 100:
        indicators += 1

    if row["Systolic_BP"] > 140:
        indicators += 1

    if row["Body_Temperature_C"] > 38:
        indicators += 1

    if row["Blood_Glucose_mg_dL"] > 180:
        indicators += 1

    if row["Health_Status"] == "Needs Attention":
        return "Needs Attention"

    elif indicators >= 2:
        return "Needs Attention"

    elif indicators == 1 and row["Health_Status"] == "Healthy":
        return "Monitor"

    else:
        return "Stable"
```

```python
patient_vitals["Health_Risk_Category"] = (
    patient_vitals.apply(
        categorize_health_risk,
        axis=1
    )
)
print(
    patient_vitals["Health_Risk_Category"]
    .value_counts()
)
```

##  Patient Monitoring Aggregations
## Group by Gender, Activity Level and Health Status

```python
patient_summary = (
    patient_vitals
    .groupby(
        [
            "Gender",
            "Activity_Level",
            "Health_Status"
        ]
    )
    .agg(
        Patient_Count=("Patient_ID", "count"),
        Average_Age=("Age", "mean"),
        Average_Heart_Rate=("Heart_Rate_bpm", "mean"),
        Average_SpO2=("SpO2_Percentage", "mean"),
        Average_Systolic_BP=("Systolic_BP", "mean"),
        Average_Temperature=("Body_Temperature_C", "mean"),
        Average_Blood_Glucose=("Blood_Glucose_mg_dL", "mean")
    )
    .reset_index()
)
```

```python
patient_summary = patient_summary.sort_values(
    "Patient_Count",
    ascending=False
)

display(patient_summary.head(10))
```

## Identify Age Groups with Higher Needs Attention

```python
age_health_summary = (
    patient_vitals
    .groupby("Age_Group")
    .agg(
        Total_Patients=("Patient_ID", "count"),
        Needs_Attention=(
            "Health_Status",
            lambda x: (x == "Needs Attention").sum()
        )
    )
    .reset_index()
)
age_health_summary["Needs_Attention_%"] = (
    age_health_summary["Needs_Attention"]
    / age_health_summary["Total_Patients"]
    * 100
)
display(
    age_health_summary.sort_values(
        "Needs_Attention_%",
        ascending=False
    )
)
```

## Activity-Level Analysis

```python
activity_summary = (
    patient_vitals
    .groupby("Activity_Level")
    .agg(
        Total_Patients=("Patient_ID", "count"),
        Needs_Attention=(
            "Health_Status",
            lambda x: (x == "Needs Attention").sum()
        )
    )
    .reset_index()
)

activity_summary["Needs_Attention_%"] = (
    activity_summary["Needs_Attention"]
    / activity_summary["Total_Patients"]
    * 100
)

display(
    activity_summary.sort_values(
        "Needs_Attention_%",
        ascending=False
    )
)
```

## Teleconsultation & Doctor Performance Analysis
## Merge Teleconsultation + Doctors

```python
tele_doctor = pd.merge(
    teleconsultation,
    doctors,
    on="Doctor_ID",
    how="left",
    suffixes=("_Consultation", "_Doctor")
)
tele_doctor_patient = pd.merge(
    tele_doctor,
    patients,
    on="Patient_ID",
    how="left",
    suffixes=("", "_Patient")
)
print(
    "Merged Shape:",
    tele_doctor_patient.shape
)

display(
    tele_doctor_patient.head()
)
```

## Doctor Performance

```python
print(tele_doctor_patient.columns.tolist())
print(doctors.columns.tolist())
```

```python
doctor_performance = (
    tele_doctor_patient
    .groupby(
        [
            "Doctor_ID",
            "Doctor_Name"
        ]
    )
    .agg(
        Consultation_Count=(
            "Consultation_ID",
            "nunique"
        ),
        Completed_Consultations=(
            "Consultation_Status",
            lambda x: (x == "Completed").sum()
        ),
        Pending_Consultations=(
            "Consultation_Status",
            lambda x: (x == "Pending").sum()
        ),
        Follow_Up_Count=(
            "Follow_Up_Date",
            lambda x: x.notna().sum()
        )
    )
    .reset_index()
)
doctor_performance["Completion_Rate_%"] = (
    doctor_performance["Completed_Consultations"]
    / doctor_performance["Consultation_Count"]
    * 100
)

doctor_performance["Follow_Up_Rate_%"] = (
    doctor_performance["Follow_Up_Count"]
    / doctor_performance["Consultation_Count"]
    * 100
)
doctor_performance = doctor_performance.sort_values(
    "Completed_Consultations",
    ascending=False
)

display(doctor_performance.head(10))

```

## Specialization Analysis

## Revenue by specialization

```python
sales_specialization = (
    sales_data
    .groupby("Specialization")
    .agg(
        Total_Revenue=("Total_Amount", "sum"),
        Average_Consultation_Amount=(
            "Total_Amount",
            "mean"
        ),
        Total_Platform_Fee=(
            "Platform_Fee",
            "sum"
        ),
        Total_Insurance_Amount=(
            "Insurance_Amount",
            "sum"
        ),
        Total_Patient_Payment=(
            "Patient_Payment",
            "sum"
        )
    )
    .reset_index()
)
display(
    sales_specialization.sort_values(
        "Total_Revenue",
        ascending=False
    )
)
```

## Revenue by Doctor

```python
doctor_revenue = (
    sales_data
    .groupby("Doctor_Name")
    .agg(
        Total_Revenue=("Total_Amount", "sum"),
        Average_Amount=("Total_Amount", "mean"),
        Patient_Payment=("Patient_Payment", "sum"),
        Insurance_Amount=("Insurance_Amount", "sum")
    )
    .reset_index()
)

display(
    doctor_revenue.sort_values(
        "Total_Revenue",
        ascending=False
    ).head(10)
)
```

## Payment Status Analysis

```python
payment_analysis = (
    sales_data
    .groupby("Payment_Status")
    .agg(
        Transaction_Count=("Sales_ID", "count"),
        Total_Revenue=("Total_Amount", "sum"),
        Patient_Payment=("Patient_Payment", "sum"),
        Insurance_Amount=("Insurance_Amount", "sum")
    )
    .reset_index()
)
total_transactions = payment_analysis[
    "Transaction_Count"
].sum()

payment_analysis["Transaction_Percentage"] = (
    payment_analysis["Transaction_Count"]
    / total_transactions
    * 100
)

display(payment_analysis)
```
